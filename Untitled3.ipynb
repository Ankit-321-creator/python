# zip function

```python
a = [1, 2, 3]
b = ['a', 'b', 'c']

zipped = zip(a, b)
print(zipped)
```

```python
a = [1, 2, 3]
b = ['a', 'b', 'c']

zipped = zip(a, b)
print(list(zipped))  # [(1, 'a'), (2, 'b'), (3, 'c')]
```

```python
a = [1, 2]
b = ['x', 'y', 'z']

print(list(zip(a, b)))  # [(1, 'x'), (2, 'y')]
```

```python
a = [10, 20, 30, 40, 50] 
b = [1, 2, 3, 4, 5] 
for a, b in zip(a, b):
    print(a , b)
```
