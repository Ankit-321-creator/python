# two sum problem

```python

list =[2,7,9,11 , 13 ,1 , 4]
target =5

for i in range(len(list)):
    for j in range (i+1 , len(list)):
        
        if list[i]+ list[j] ==target:
         print(list[i] , list[j])
         break
        
```

# print duplicate value form a list

```python
list =[1,2,3, 4 ,4,1,2 ,4 ,3 , 5,6 ,7]
num  =[]
for i in range(0 , len(list)):
    for j in range(i+1 , len(list)):
        if list[i]== list[j] and list[i] not in num
         num.append(list[i])
print(num)
```

# Reversing  element from a list

```python
list =[1,2 ,3 ,4 ,5]
list.reverse()
list

```

```python
list =[1,2 ,3 ,4 ,5]
left =0
right =len(list)-1
while left < right :
    
    list[left] , list[right] =list[right] , list[left]
    left +=1
    right-=1

print(list)

```
